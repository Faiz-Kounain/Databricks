```python
# Read the task values from Task 1
row_count = dbutils.jobs.taskValues.get(
    taskKey="write_data",     # name of the producer task
    key="row_count",          # value to fetch
    default=0                 # fallback if missing
)

status = dbutils.jobs.taskValues.get(
    taskKey="write_data",
    key="status",
    default="UNKNOWN"
)

print(f"📩 Received from Task 1: row_count={row_count}, status={status}")

# Use the value to make a decision
if row_count == 0:
    print(" No data was written. Skipping further processing.")
    dbutils.notebook.exit("skipped_due_to_empty_data")
```
